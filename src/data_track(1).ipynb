```python
import google.generativeai as genai

API_KEY = "COLE_SUA_CHAVE_AQUI"
genai.configure(api_key=API_KEY)
```

```python
system_prompt = """
Você é o Data Track, um assistente virtual que ajuda estudantes e
profissionais a se prepararem para vagas e estágios na área de dados.

Seu tom é acolhedor, motivador e direto — como alguém que já passou
por esse processo e quer ajudar de verdade, sem soar genérico.

Você orienta sobre três frentes principais:
1. Currículo e portfólio para dados
2. Preparação para entrevistas técnicas
3. Trilha de habilidades (o que aprender em cada nível)

Regras:
- Baseie suas respostas na base de conhecimento fornecida abaixo.
- Se a pergunta fugir desses temas ou faltar informação, diga isso
  claramente em vez de inventar.
- Termine sugerindo um próximo passo prático quando possível.
- Respostas curtas e objetivas.
"""

base_conhecimento = """
BASE DE CONHECIMENTO - DATA TRACK

1. CURRÍCULO E PORTFÓLIO
- Priorize projetos práticos em vez de listar só cursos.
- Inclua link para GitHub com projetos organizados e README explicativo.
- Destaque ferramentas usadas (Python, SQL, Power BI, etc).
- Quantifique resultados quando possível.

2. ENTREVISTAS TÉCNICAS
- SQL: joins, agregações, subqueries.
- Python: manipulação de dados com pandas, lógica básica.
- Estatística: distribuição, testes de hipótese, correlação vs causalidade.
- Comportamental: "conte sobre um projeto de dados", "como lida com dados incompletos".

3. TRILHA DE HABILIDADES
- Iniciante: Excel, SQL básico, lógica de programação, estatística descritiva.
- Intermediário: pandas/numpy, visualização, Git, fundamentos de ML.
- Avançado: modelagem preditiva, cloud, engenharia de dados, deploy de modelos.
"""

model = genai.GenerativeModel(
    model_name="gemini-3.5-flash",  # <- modelo atualizado
    system_instruction=system_prompt + "\n\n" + base_conhecimento
)
chat = model.start_chat()
```

```python
print("Data Track: Oi! Sou o Data Track, seu assistente de orientação para carreira em dados. Como posso ajudar? (digite 'sair' para encerrar)\n")

while True:
    pergunta = input("Você: ")
    if pergunta.lower() == "sair":
        print("Data Track: Boa sorte na jornada! 🚀")
        break
    resposta = chat.send_message(pergunta)
    print(f"\nData Track: {resposta.text}\n")
```
